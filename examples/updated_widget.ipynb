```python
import asyncio    
from ipylgbst import LegoBoostWidget,LedColor,Port, Sensor
from ipywidgets import Output
from IPython.display import display

# the robot's LED supports the following colors
colors = [
    LedColor.pink,
    LedColor.purple,
    LedColor.blue,
    LedColor.lightblue,
    LedColor.cyan,
    LedColor.green,
    LedColor.yellow,
    LedColor.orange,
    LedColor.red,
]
```

# Connect the device 
Note that connecting the device should take 5-10 sec

```python
boost = LegoBoostWidget()
task1 = boost.connect()
boost
```

```python
task2 = boost.set_led("orange")
```

```python
task3 = await boost.set_led_async("green")
```

# Change Color of LED

```python
for color in colors:
    print(f"set color to {color.value}")
    await boost.set_led_async(color)
    await asyncio.sleep(0.5)
```
